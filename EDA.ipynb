```python
import pandas as pd

df_act = pd.read_csv("./Data/raw/Censo-local-historico/2021-12/209548-711-censo-actividades-historico-csv.csv", sep=';')
print(f"Filas: {len(df_act)}")
print(f"Columnas: {df_act.columns.tolist()}")
print(f"Distritos únicos: {df_act['desc_distrito_local'].unique()}")
print(f"Barrios únicos: {df_act['desc_barrio_local'].unique()}")
print(f"\nPriumeras 3 filas:")
df_act.head(3)
```

```python
# ¿ Que estados  hay?
print("=" * 60)
print("ESTADOS DE LOCALES (desc_situacion_local):")
print("=" * 60)
print(df_act['desc_situacion_local'].value_counts())

# ¿Qué actividades hay?
print("=" * 60)
print("ACTIVIDADES ECONÓMICAS ÚNICAS (desc_epigrafe):")
print("=" * 60)
print(F"Total de actividades diferentes: {df_act['desc_epigrafe'].nunique()}")
print(f"\nPrimeras 20")
print(df_act['desc_epigrafe'].unique()[:20])

# ¿Cuáles contiene BAR, CAFÉ, RESTAURANTE?
print("=" * 60)
print("ACTIVIDADES CON BAR, CAFÉ, RESTAURATE :")
print("=" * 60)
hosteleria = df_act[df_act['desc_epigrafe'].str.contains('BAR|CAFÉ|RESTAURANTE', case=False, na=False)]
print(f"Locales de hosteleria: {len(hosteleria)}")
print(hosteleria['desc_epigrafe'].nunique())

# Ejemplo: un barrio especifico
print("\n" + "=" * 60)
print("EJEMPLO: BARRIO 'MALASAÑA' (o similar):")
print("=" * 60)
malasana = df_act[df_act['desc_barrio_local'].str.contains('MALASAÑA|TRIBUNAL|ATOCHA', case=False, na=False)]
print(f"Locales en barrio tipo: {len(malasana)}")
if len(malasana) > 0:
    print(malasana[['desc_barrio_local', 'desc_epigrafe', 'desc_situacion_local']].head(10))
```

```python
actividades_hosteleria = hosteleria['desc_epigrafe'].unique()
print("CATEGORÍAS DE HOSTELERÍA (17 tipos)")
print("=" *80)

for i, actividad in enumerate(sorted(actividades_hosteleria), 1):
    count = len(hosteleria[hosteleria['desc_epigrafe'] == actividad])
    print(f"{i:2d}. [{count:5d}] {actividad}")
```

```python
df_act = pd.read_csv("./Data/raw/Censo-local-historico/2021-12/209548-711-censo-actividades-historico-csv.csv", sep=';')
df_loc = pd.read_csv("./Data/raw/Censo-local-historico/2021-12/209548-277-censo-locales-historico-csv.csv", sep=';')

# ¿Quantos locales hay únicos en cada archivo?
print(f"Locales únicos en LOCALES: {df_loc['id_local'].nunique()}")
print(f"Locales únicos en ACTIVIDADES: {df_act['id_local'].nunique()}")

# ¿Hay locales en LOCALES uqe no aparecen en ACTIVIDADES?
locales_sin_actividad = df_loc[~df_loc['id_local'].isin(df_act['id_local'])]
print(f"\n Locales sin actividad registrada: {len(locales_sin_actividad)}")

# Puede un local tener multiples actividades
actividades_por_local = df_act.groupby('id_local').size()
print(f"\n Locales con múltiples actividades: {(actividades_por_local > 1).sum()}")
print(f"Máximo de actividesdes en un local: {actividades_por_local.max()}")
```
